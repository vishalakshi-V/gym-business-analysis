```python
import pandas as pd
import matplotlib.pyplot as plt

print("loaded ✅")
```

```python
revenue_df = pd.read_excel("gym_data.xlsx", sheet_name="Revenue")
print(revenue_df)
```

```python
plt.figure(figsize=(12, 5))
plt.plot(revenue_df["month"], revenue_df["total_revenue"], marker="o", color="blue", linewidth=2)
plt.title("Monthly Total Revenue - 2023", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
labels = ["Membership Revenue", "Personal Training", "Supplement Sales"]
values = [
    revenue_df["membership_revenue"].sum(),
    revenue_df["personal_training_revenue"].sum(),
    revenue_df["supplement_sales"].sum()
]

plt.figure(figsize=(7, 7))
plt.pie(values, labels=labels, autopct="%1.1f%%", colors=["#2ecc71", "#3498db", "#e74c3c"], startangle=140)
plt.title("Revenue Split by Source - 2023", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
members_df = pd.read_excel("gym_data.xlsx", sheet_name="Members")

revenue_by_type = members_df.groupby("membership_type")["monthly_fee"].sum()

plt.figure(figsize=(8, 5))
plt.bar(revenue_by_type.index, revenue_by_type.values, color=["#3498db", "#e74c3c", "#2ecc71"])
plt.title("Total Revenue by Membership Type", fontsize=16)
plt.xlabel("Membership Type")
plt.ylabel("Total Revenue (₹)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.bar(revenue_df["month"], revenue_df["membership_revenue"], label="Membership", color="#3498db")
plt.bar(revenue_df["month"], revenue_df["personal_training_revenue"], bottom=revenue_df["membership_revenue"], label="Personal Training", color="#2ecc71")
plt.bar(revenue_df["month"], revenue_df["supplement_sales"], 
        bottom=revenue_df["membership_revenue"] + revenue_df["personal_training_revenue"], 
        label="Supplements", color="#e74c3c")

plt.title("Monthly Revenue Breakdown - 2023", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

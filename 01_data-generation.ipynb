```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

print("Libraries loaded ✅")
```

```python
np.random.seed(42)

names = [
    "Ravi Kumar", "Priya S", "Anjali R", "Suresh M", "Deepa K",
    "Kiran B", "Sneha T", "Arjun P", "Meera N", "Vikram S",
    "Pooja H", "Rahul G", "Divya L", "Arun J", "Lakshmi V",
    "Manoj C", "Swathi D", "Naveen R", "Kavya M", "Prasad B",
    "Nisha K", "Rohit A", "Suma P", "Ganesh T", "Rekha S",
    "Amit N", "Usha R", "Sanjay M", "Bindhu K", "Harish V"
]

all_names = (names * 5)[:150]
member_ids = [f"M{str(i+1).zfill(3)}" for i in range(150)]
membership_types = np.random.choice(["Basic", "Premium", "Student"], 150, p=[0.5, 0.35, 0.15])
monthly_fees = {"Basic": 800, "Premium": 1500, "Student": 500}
fees = [monthly_fees[m] for m in membership_types]

start_date = datetime(2023, 1, 1)
join_dates = [start_date + timedelta(days=random.randint(0, 364)) for _ in range(150)]
statuses = np.random.choice(["Active", "Inactive"], 150, p=[0.72, 0.28])

members_df = pd.DataFrame({
    "member_id": member_ids,
    "name": all_names,
    "join_date": join_dates,
    "membership_type": membership_types,
    "monthly_fee": fees,
    "status": statuses
})

print(members_df.head())
print(f"\nTotal members: {len(members_df)}")
```

```python
months = pd.date_range(start="2023-01-01", periods=12, freq="MS")
month_labels = [m.strftime("%b-%Y") for m in months]

membership_revenue = np.random.randint(40000, 80000, 12)
pt_revenue = np.random.randint(8000, 20000, 12)
supplement_sales = np.random.randint(3000, 10000, 12)
total_revenue = membership_revenue + pt_revenue + supplement_sales

revenue_df = pd.DataFrame({
    "month": month_labels,
    "membership_revenue": membership_revenue,
    "personal_training_revenue": pt_revenue,
    "supplement_sales": supplement_sales,
    "total_revenue": total_revenue
})

print(revenue_df)
```

```python
attendance_records = []
time_slots = ["Morning", "Afternoon", "Evening"]
slot_probs = [0.45, 0.2, 0.35]

for _, member in members_df.iterrows():
    visits = random.randint(5, 80)
    for _ in range(visits):
        visit_date = start_date + timedelta(days=random.randint(0, 364))
        attendance_records.append({
            "member_id": member["member_id"],
            "visit_date": visit_date,
            "time_slot": np.random.choice(time_slots, p=slot_probs),
            "day_of_week": visit_date.strftime("%A")
        })

attendance_df = pd.DataFrame(attendance_records)
print(f"Total visit records: {len(attendance_df)}")
print(attendance_df.head())
```

```python
with pd.ExcelWriter("gym_data.xlsx", engine="openpyxl") as writer:
    members_df.to_excel(writer, sheet_name="Members", index=False)
    revenue_df.to_excel(writer, sheet_name="Revenue", index=False)
    attendance_df.to_excel(writer, sheet_name="Attendance", index=False)

print("gym_data.xlsx saved successfully! ✅")
```
